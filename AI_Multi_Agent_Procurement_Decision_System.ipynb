# AI Multi-Agent Procurement Decision System

A multi-agent AI system that compares vendors using deterministic weighted scoring, AI-powered procurement analysis, vendor risk assessment, recommendation generation, and human approval.

## Key Features

- Multi-vendor comparison
- Weighted decision scoring
- Price normalization
- Delivery speed scoring
- Quality evaluation
- Reliability evaluation
- AI procurement analysis
- Vendor risk assessment
- AI recommendation
- Human approval
- Final structured procurement report

## How It Works

1. Vendor offers are provided with price, quality, delivery time, reliability, and notes.
2. Python normalizes price and delivery performance.
3. The Python Scoring Engine calculates a weighted score for every vendor.
4. Vendors are ranked using verified numerical results.
5. The Procurement Analyst Agent explains the main trade-offs.
6. The Vendor Risk Agent identifies potential procurement risks.
7. The Recommendation Agent recommends the strongest option.
8. A human approves or rejects the recommendation.
9. The system generates a final procurement decision report.

## Decision Weights

```text
Price        35%
Quality      30%
Delivery     20%
Reliability  15%
```

## Weighted Scoring

The final score is calculated using:

```text
Final Score =
Price Score × 35%
+
Quality Score × 30%
+
Delivery Score × 20%
+
Reliability Score × 15%
```

Python performs the calculations instead of relying on the AI model.

````python
!pip install -q groq

from groq import Groq
from google.colab import userdata

GROQ_API_KEY = userdata.get("GROQ_API_KEY")

client = Groq(api_key=GROQ_API_KEY)

MODEL = "qwen/qwen3.6-27b"

print("Procurement Decision AI ready ✅")
````

````python
# ==========================================================
# PROJECT 9 - MULTI-AGENT PROCUREMENT DECISION SYSTEM
# ==========================================================


# ----------------------------------------------------------
# AI HELPER
# ----------------------------------------------------------

def ask_ai(prompt):

    response = client.chat.completions.create(
        model=MODEL,
        messages=[
            {
                "role": "user",
                "content": prompt
            }
        ]
    )

    return response.choices[0].message.content


# ----------------------------------------------------------
# SAMPLE VENDORS
# ----------------------------------------------------------

vendors = [
    {
        "name": "Vendor A",
        "price": 85000,
        "quality_score": 92,
        "delivery_days": 12,
        "reliability_score": 90,
        "notes": "Strong reputation and consistent delivery history."
    },

    {
        "name": "Vendor B",
        "price": 72000,
        "quality_score": 78,
        "delivery_days": 8,
        "reliability_score": 72,
        "notes": "Lowest price but some previous delivery inconsistencies."
    },

    {
        "name": "Vendor C",
        "price": 79000,
        "quality_score": 88,
        "delivery_days": 6,
        "reliability_score": 85,
        "notes": "Good quality and fastest delivery with positive history."
    }
]


# ----------------------------------------------------------
# DECISION WEIGHTS
# ----------------------------------------------------------

weights = {
    "price": 0.35,
    "quality": 0.30,
    "delivery": 0.20,
    "reliability": 0.15
}


# ----------------------------------------------------------
# 1. PYTHON WEIGHTED SCORING ENGINE
# ----------------------------------------------------------

def score_vendors(vendors, weights):

    prices = [v["price"] for v in vendors]
    delivery_days = [v["delivery_days"] for v in vendors]

    lowest_price = min(prices)
    fastest_delivery = min(delivery_days)

    scored = []

    for vendor in vendors:

        # Lower price = better score
        price_score = (
            lowest_price / vendor["price"]
        ) * 100

        # Faster delivery = better score
        delivery_score = (
            fastest_delivery / vendor["delivery_days"]
        ) * 100

        final_score = (
            price_score * weights["price"]
            +
            vendor["quality_score"] * weights["quality"]
            +
            delivery_score * weights["delivery"]
            +
            vendor["reliability_score"] * weights["reliability"]
        )

        scored.append({
            **vendor,
            "price_score": round(price_score, 2),
            "delivery_score": round(delivery_score, 2),
            "final_score": round(final_score, 2)
        })

    return sorted(
        scored,
        key=lambda x: x["final_score"],
        reverse=True
    )


vendor_scores = score_vendors(
    vendors,
    weights
)


# ----------------------------------------------------------
# 2. PROCUREMENT ANALYST AGENT
# ----------------------------------------------------------

def procurement_analyst_agent(vendor_scores):

    prompt = f"""
You are a procurement analysis agent.

Here are vendor scores calculated by Python:

{vendor_scores}

Analyze the vendors.

Explain:

- Cost position
- Quality
- Delivery speed
- Reliability
- Main trade-offs

Do NOT change or invent numerical scores.

Return:

PROCUREMENT ANALYSIS

Vendor A:
...

Vendor B:
...

Vendor C:
...

KEY TRADE-OFFS:
...
"""

    return ask_ai(prompt)


procurement_analysis = procurement_analyst_agent(
    vendor_scores
)


# ----------------------------------------------------------
# 3. RISK AGENT
# ----------------------------------------------------------

def vendor_risk_agent(vendor_scores):

    prompt = f"""
You are a vendor risk analysis agent.

Vendor information:

{vendor_scores}

Identify realistic procurement risks for each vendor.

Use only the provided information.

For each vendor return:

VENDOR:
...

RISKS:
- ...

RISK LEVEL:
LOW / MEDIUM / HIGH

Do not invent facts.
"""

    return ask_ai(prompt)


risk_analysis = vendor_risk_agent(
    vendor_scores
)


# ----------------------------------------------------------
# 4. RECOMMENDATION AGENT
# ----------------------------------------------------------

def recommendation_agent(
    vendor_scores,
    procurement_analysis,
    risk_analysis
):

    prompt = f"""
You are a senior procurement decision advisor.

Verified Python Scores:
{vendor_scores}

Procurement Analysis:
{procurement_analysis}

Risk Analysis:
{risk_analysis}

Recommend the best vendor.

Rules:

- Give strong weight to the verified Python ranking.
- Consider vendor risk.
- Explain important trade-offs.
- Do not invent information.
- Do not make the final purchasing decision.

Return:

RECOMMENDED VENDOR:
...

REASON:
...

MAIN TRADE-OFF:
...

ALTERNATIVE:
...
"""

    return ask_ai(prompt)


recommendation = recommendation_agent(
    vendor_scores,
    procurement_analysis,
    risk_analysis
)


# ----------------------------------------------------------
# SHOW RANKING
# ----------------------------------------------------------

print("\n============================")
print("VENDOR RANKING")
print("============================")

for index, vendor in enumerate(
    vendor_scores,
    start=1
):

    print(
        f"{index}. {vendor['name']} "
        f"- Score: {vendor['final_score']}/100"
    )


print("\n============================")
print("AI RECOMMENDATION")
print("============================")

print(recommendation)


# ----------------------------------------------------------
# 5. HUMAN APPROVAL
# ----------------------------------------------------------

print("\n============================")
print("HUMAN APPROVAL REQUIRED")
print("============================")

human_decision = input(
    "Type APPROVE or REJECT: "
).strip().upper()


if human_decision == "APPROVE":

    approval_status = "APPROVED ✅"

else:

    approval_status = "REJECTED ❌"


# ----------------------------------------------------------
# 6. FINAL PROCUREMENT REPORT
# ----------------------------------------------------------

final_report = f"""
=================================================
AI MULTI-AGENT PROCUREMENT DECISION REPORT
=================================================

DECISION WEIGHTS
----------------
Price: 35%
Quality: 30%
Delivery: 20%
Reliability: 15%


VERIFIED VENDOR RANKING
-----------------------
{vendor_scores}


PROCUREMENT ANALYSIS
--------------------
{procurement_analysis}


VENDOR RISK ANALYSIS
--------------------
{risk_analysis}


AI RECOMMENDATION
-----------------
{recommendation}


HUMAN DECISION
--------------
{approval_status}


=================================================
END OF REPORT
=================================================
"""

print("\n")
print(final_report)
````
